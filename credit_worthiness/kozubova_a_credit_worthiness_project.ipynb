# Исследование надежности заемщиков 

**Цель исследования** — разобраться, какие факторы влияют на факт погашения кредита в срок.

**Для этого проверим 4 гипотезы**:
1. Количество детей влияет на возврат кредита в срок.
2. Семейное положение клиента связано с возвратом кредита в срок.
3. Уровень дохода влияет на вероятность погашения кредита вовремя.
4. Разные цели кредита оказывают влияние на его возврат в срок.

**Описание данных**

В этом исследовании будут использованы входные данные от банка — статистика о платёжеспособности клиентов. Мы получим их из файла `credit_worthiness_project`. Данные включают следующую информацию о клиентах банка:

* `family_status_id` — идентификатор семейного положения


* `gender` — пол клиента


* `income_type` — тип занятости


* `debt` — имел ли задолженность по возврату кредитов


* `total_income` — ежемесячный доход в рублях


* `purpose` — цель получения кредита

## Предобработка данных

### Удаление пропусков

```python
# выведем количество пропущенных значений для каждого столбца, используем комбинацию двух методов
data.isna().sum()  
```

В двух столбцах есть пропущенные значения. Один из них — `days_employed`. Пропуски в этом столбце обработаем на следующем этапе. Другой столбец с пропущенными значениями — `total_income` — хранит данные о доходах. На сумму дохода сильнее всего влияет тип занятости, поэтому заполнить пропуски в этом столбце нужно медианным значением по каждому типу из столбца `income_type`. Например, у человека с типом занятости сотрудник пропуск в столбце `total_income` должен быть заполнен медианным доходом среди всех записей с тем же типом.

```python
# найдём медиану дохода по каждому типу деятельности из столбца income_type
# заполним пропуски медианой в столбце total_income и сохраним результат
data['total_income'] = data['total_income'].fillna(data.groupby('income_type')['total_income'].transform('median')) 

# убедимся, что пропуски заполнены 
data.isna().sum()
```

### Обработка аномальных значений

Обработаем значения в столбце `days_employed`: заменим все отрицательные значения положительными: 

```python
# заменим значения с помощью метода abs() и сохраним результат 
data['days_employed'] = data['days_employed'].abs() 

# выведем первые 20 строк датафрейма и проверим результат
data.head(10) 
```

Для каждого типа занятости выведем медианное значение трудового стажа `days_employed` в днях:

```python
# найдём медиану трудового стажа по каждому типу деятельности из столбца income_type
data.groupby('income_type')['days_employed'].median() 
```

У двух типов (безработные и пенсионеры) получатся аномально большие значения. Исправить такие значения сложно, поэтому оставим их как есть. Тем более этот столбец не понадобится нам для исследования.

Проверим столбцы `children`, `debt` и `family_status` на наличие аномальных значений.

Выведем перечень уникальных значений столбца `children`:

```python
# используем метод unique() для получения уникальных значений
data['children'].unique()
```

В столбце `children` есть два аномальных значения: -1 и 20. Удалим строки, в которых встречаются такие аномальные значения из датафрейма data.

```python
# удалим аномальные значения и сохраним результат в data
data = data[(data['children'] != -1) & (data['children'] != 20)]

# ещё раз выведем список уникальных значений для проверки результата
data['children'].unique()
```

Выведем на экран перечень уникальных значений столбцов `debt` и `family_status`:

Убедились, что в столбцах `debt` и `family_status` нет аномальных значений.

### Удаление пропусков (продолжение)

Заполним пропуски в столбце `days_employed` медианными значениями по каждому типу занятости `income_type`:

```python
# найдём медиану трудового стажа по каждому типу деятельности из столбца income_type
# заполним пропуски медианой в столбце days_employed и сохраним результат
data['days_employed'] = data['days_employed'].fillna(data.groupby('income_type')['days_employed'].transform('median')) 

# убедимся, что пропуски заполнены 
data.isna().sum()
```

### Изменение типов данных

В этом исследовании сосредоточимся на доходе в рублях без копеек. Целые числа легче интерпретировать и сравнивать, поэтому заменим вещественный тип данных в столбце `total_income` на целочисленный:

Обработаем неявные дубликаты в столбце `education`:

```python
# приведём значения к нижнему регистру с помощью метода str.lower()
data['education'] = data['education'].str.lower()

# выведем 20 строк датафрейма и проверим результат
data.head(20)
```

Проверим остальные столбцы на наличие дубликатов:

### Категоризация данных

На основании диапазонов, указанных ниже, создадим в датафрейме `data` столбец `total_income_category` с категориями:

- 0–30000 — `'E'`;
- 30001–50000 — `'D'`;
- 50001–200000 — `'C'`;
- 200001–1000000 — `'B'`;
- 1000001 и выше — `'A'`.


Например, кредитополучателю с доходом 25000 назначим категорию `'E'`, а клиенту, получающему 235000, — `'B'`. Используем собственную функцию с именем `categorize_income()` и метод `apply()`.

```python
# напишем функцию для разделения данных на категории с параметром - столбец с ежемесячным доходом клиента
# в теле функции используем условные конструкции для разделения данных
def categorize_income(total_income): 
    
    if 0 <= total_income <= 30000:
        return 'E'
    if 30001 <= total_income <= 50000:
        return 'D'
    if 50001 <= total_income <= 200000:
        return 'C'
    if 200001 <= total_income <= 1000000:
        return 'B'
    return 'A'

# выведем результат, чтобы протестировать функцию
# print(categorize_income(50005))
```

```python
# применим функцию методом apply() и сохраним результат в столбец total_income_category
data['total_income_category'] = data['total_income'].apply(categorize_income) 

# проверим результат
data.head()
```

Создадим в датафрейме `data` столбец `age_category` с категориями на основании следующих диапазонов:

- 18–25 — `от 18 до 25 лет`;
- 25-30 — `от 25 до 30 лет`;
- 30-40 — `от 30 до 40 лет`;
- 40-50 — `от 40 до 50 лет`;
- 50 и выше — `от 50 лет`.


Например, клиенту в возрасте 20 лет назначим категорию ``от 18 до 25 лет``. Напишем собственную функцию с именем `categorize_age` и используем метод `apply()`.

```python
# напишем функцию для разделения данных на категории с параметром - столбец с возрастом клиента
# в теле функции используем условные конструкции для разделения данных
def categorize_age(age):
    
    if 18 <= age <= 25:
        return 'от 18 до 25 лет'
    elif 25 <= age <= 30:
        return 'от 25 до 30 лет'
    elif 30 <= age <= 40:
        return 'от 30 до 40 лет'
    elif 40 <= age <= 50:
        return 'от 40 до 50 лет'
    return 'от 50 лет'

# выведем результат, чтобы протестировать функцию
# print(categorize_age(19))
```

```python
# применим функцию методом apply() и сохраним результат в столбец age_category
data['age_category'] = data['dob_years'].apply(categorize_age)

# проверим результат
data.head()
```

Перед тем, как разделить данные на категории по целям кредита, выведем на экран перечень уникальных целей взятия кредита из столбца `purpose`:

```python
# выведем список уникальных значений с помощью метода unique()
data['purpose'].unique()
```

Создадим функцию, которая на основании данных из столбца `purpose` сформирует новый столбец `purpose_category`, в который войдут следующие категории:

- `'операции с автомобилем'`;
- `'операции с недвижимостью'`;
- `'проведение свадьбы'`;
- `'получение образования'`.

Например, если в столбце `purpose` находится подстрока `'на покупку автомобиля'`, то в столбце `purpose_category` должна появиться строка `'операции с автомобилем'`.

Используем собственную функцию с именем `categorize_purpose()` и метод `apply()`:

```python
# напишем функцию для разделения данных на категории с параметром - строка в датафрейме
# в теле функции используем условные конструкции для разделения данных
def categorize_purpose(row):
    if 'автом' in row:
        return 'операции с автомобилем'
    elif 'жил' in row or 'недвиж' in row:
        return 'операции с недвижимостью'
    elif 'свад' in row:
        return 'проведение свадьбы'
    elif 'образов' in row:
        return 'получение образования'
    return 'нет категории'

# выведем результат, чтобы протестировать функцию
# print(categorize_purpose('свадьба')) 
```

```python
# применим функцию методом apply() и сохраним результат в столбец purpose_category
data['purpose_category'] = data['purpose'].apply(categorize_purpose) 

# проверим результат
data.head(10)
```

В случае с количественными переменными лучше заполнить пропуски медианным значением, так как в данных возможны значения, которые значительно отличаются от других и поэтому могут сместить среднее. 

Отрицательные значения в столбце `days_employed` были заменены на положительные. Устранены артефакты в стобце `children`.

Вещественный тип данных в столбце `total_income` заменили на целочисленный. Удалили дубликаты в данных. Разбили данные в столбцах `total_income`, `dob_years` и `purpose` на категории и сохранили в новых столбцах `total_income_category`, `age_category` и `purpose_category`.

## Проверка гипотез

###  Анализ взаимосвязи между количеством детей и возвратом кредита в срок

Чтобы проверить, влияет ли количество детей клиента на факт погашения кредита в срок: 

* Посмотрим, сколько клиентов входит в каждую категорию по количеству детей.
* Проверим, сколько клиентов в каждой группе имеют задолженность по возврату кредитов.
* Выясним, какую долю от всех клиентов составляют клиенты с задолженностью в каждой категории.

```python
# используем методы groupby() и agg() вместе, чтобы одновременно применить две функции к столбцу 'children'
# выведем на экран общее количество клиентов и количество клиентов с задолженностью в каждой категории столбца 'children'
data.groupby('children').agg({'debt':['count', 'sum']})
```

Наибольшее число клиентов, представленных в таблице, `не имеют детей`. Далее следуют категории `с одним ребенком` и `двумя детьми`. Эти же категории охватывают самое большое число клиентов с задолженностью.

Самая маленькая группа состоит из клиентов с количеством детей - 5. Согласно данным, все клиенты в категории `"5"` возвращают кредит в срок.

Чтобы найти долю клиентов с задолженностью рассчитаем конверсию. Разделим количество клиентов с задолженностью на общее количество клиентов во всех категориях:

```python
# объявим функцию для рассчёта конверсии с параметрами: таблица с данными, столбцы категории клиентов и залдолженности 
def conversion(data, category, debt): 

# сгруппируем данные по столбцу с категорией и применим к нему агрегирующие функции
# рассчитаем долю клиентов с задолженностью и округлим результат
# вернём результат, отсортированный в порядке убывания
    category_count_sum = data.groupby(category)[debt].agg(['count', 'sum'])
    category_count_sum['conversion'] = (category_count_sum['sum'] / category_count_sum['count'] * 100).round(1)
    return category_count_sum['conversion'].sort_values(ascending=False)
```

```python
# вызовем функцию, передадим ей таблицу с данными, столбцы 'children' и 'debt' в качестве аргументов
# выведем результат на экран
children_conversion = conversion(data, 'children', 'debt')
children_conversion
```

Соберём полученные данные в таблицу для более комфортного восприятия:

```python
# создадим таблицу с помощью конструктора библиотеки pandas
# сохраним таблицу в переменную children_conversion_table
children_conversion_table = pd.DataFrame(  
    data=[
        [4, 41, 4, 9.8], 
        [2, 2052, 194, 9.5], 
        [1, 4808, 444, 9.2], 
        [3, 330, 27, 8.2], 
        [0, 14091, 1063, 7.5],
        [5, 9, 0, 0.0]
        
    ],
    columns = ['children_category', 'customers_number', 'debt', 'conversion_%']
)

# выведем результат на экран
children_conversion_table
```

**Вывод:** 

В результате расчетов выяснили:
1. Большинство клиентов с задолженностью входит в категорию `c 4 детьми` - **9.8%**. В ней задолженность имеет каждый десятый клиент.
2. Близкое значение имеют группы: `с двумя детьми` - **9.5%**, `с одним ребёнком` - **9.2%**.
2. Меньше доля клиентов с задолженностью в категориях `с тремя детьми` и `без детей` - **8.2%** и **7.5%**.
3. Клиенты `с 5 детьми` **не имеют задолженностей** по кредиту. В эту категорию входит самое маленькое количество клиентов - всего лишь 9 человек.

Данные показывают, что нельзя утверждать о прямой зависимости между количеством детей и возвратом кредита в срок, так как у клиентов `с 1 ребёнком` задолженность по кредиту на **1%** выше, чем у клиентов `с тремя детьми`. С другой стороны, клиенты `без детей` имеют самый низкий процент задолженности. `Пятая категория` включает слишком маленькое количество клиентов для корректного вывода. Таким образом, есть слабая зависимость между количеством детей и погашением кредита в срок.

### Анализ взаимосвязи между целью кредита и возвратом кредита в срок

Выясним, как разные цели кредита влияют на его оплату в срок:

* Найдём общее количество клиентов и количество клиентов с задолженностью в каждой категории по цели кредита.
* Рассчитаем долю клиентов с задолженностью.
* Для каждой цели кредита выявим долю клиентов с задолженностью в зависимости от уровня дохода, семейного положения и возраста клиентов.

```python
# применим функции count и sum к столбцу 'purpose_category' и ознакомимся с результатом
data.groupby('purpose_category').agg({'debt': ['count', 'sum']}).sort_values(by=[('debt','count'), ('debt','sum')], ascending=False) 
```

Самая большая группа по количеству клиентов в целом и количеству клиентов с задолженностью - `"операции с недвижимостью"`, самая маленькая группа - `"проведение свадьбы"`.	

Выявим долю клиентов с задолженностью для каждой цели кредита:

```python
# вызовем функцию для расчёта конверсии, передадим ей таблицу с данными, столбцы 'purpose_category' и 'debt'
# выведем результат на экран
purpose_category_conversion = conversion(data, 'purpose_category', 'debt')
purpose_category_conversion
```

Представим результат в виде таблицы:

```python
# создадим таблицу и сохраним её в переменную purpose_conversion_table
purpose_conversion_table = pd.DataFrame(  
    data=[
        ["операции с автомобилем ", 4279, 400, 9.3], 
        ["получение образования ", 3988, 369, 9.3], 
        ["проведение свадьбы", 2313, 183, 7.9], 
        ["операции с недвижимостью", 10751, 780, 7.3]
        
    ],
    columns = ['purpose_category', 'customers_number', 'debt', 'conversion_%']
)

# выведем результат на экран
purpose_conversion_table
```

**Посмотрим, как разные цели кредита влияют на его возврат в срок. Для этого рассчитаем долю клиентов с задолженностью для каждой цели кредита:**

* `операции с автомобилем`
* `получение образования`
* `проведение свадьбы`
* `операции с недвижимостью`

**В зависимости от категорий**:

* `уровень дохода`

* `семейное положение`

* `возраст клиентов`

Исключили категорию `количество детей` клиента, так как выявили, что она практически не влияет на возврат кредита в срок.

Рассчитаем долю клиентов с задолженностью с целью `"операции с автомобилем"` в зависимости от `ежемесячного дохода`:

```python
# напишем функцию для рассчёта конверсии с параметрами: таблица с данными, столбец с целью кредита, столбец с категорией
# выберем строки с целью кредита в столбце 'purpose_category' и отфильтруем данные, сохраним результат
# сгруппируем отфильтрованные данные по столбцам с целью кредита и категорией
# посчитаем количество клиентов и количество клиентов с задолженностью  
# сбросим индекс
def category_conversion(data, purpose_category, category_column): 
    purpose_category_data = data[data['purpose_category'] == purpose_category]
    purpose_count_sum = purpose_category_data.groupby(['purpose_category', category_column]).agg(
    customer_count=(category_column,'count'),
    debt_sum=('debt','sum')
    ).reset_index()

# рассчитаем конверсию, округлим результат и добавим новый столбец в таблицу
# отсортируем значения по убыванию и заново пронумеруем строки
# вернём таблицу с результатом

    purpose_count_sum['conversion_%'] = round(
        (purpose_count_sum['debt_sum'] / purpose_count_sum['customer_count'] * 100), 1
    )
    purpose_count_sum = purpose_count_sum.sort_values(by='conversion_%', ascending=False).reset_index(drop=True)
    return purpose_count_sum
```

```python
# выведем на экран результат для цели "операции с автомобилем" и "категории ежемесячного дохода"
category_conversion(data, 'операции с автомобилем', 'total_income_category')
```

Самая высокая доля клиентов с задолженностью с целью `"операции с автомобилем"` в зависимости от `ежемесячного дохода` находится в категории `"D"`-  13.9%. Клиенты в этой категории имеют доход `от 30 001 до 50 000`. Клиенты с целью операций с автомобилем в категориях дохода `"A" - от 1 000 001` и `"E" - 0-30 000` не имеют задолженностей.

Рассчитаем долю клиентов с задолженностью с целью `"операции с автомобилем"` в зависимости от `семейного положения`:

```python
# выведем на экран результат для цели "операции с автомобилем" и "категории семейного положения"
category_conversion(data, 'операции с автомобилем', 'family_status')
```

Самая высокая доля клиентов с задолженностью с целью `"операции с автомобилем"` -  12.9%. Клиенты в этой категории `не женаты / не замужем`. Далее следует категория клиентов с задолженностью `в гражданском браке` -  11.9%. Самая низкая доля у клиентов `в разводе` - 7.5%.
