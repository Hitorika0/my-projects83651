# Исследование данных о российском кинопрокате

**Цель исследования** — выявить текущие тренды на рынке российского кинопроката. 

**Для этого**:

* Изучим фильмы, которые вышли в прокат в России.
* Уделим внимание фильмам, которые получили государственную поддержку.
* Исследуем, насколько такие фильмы интересны зрителю.

**Описание данных**

В этом исследовании будут использованы данные, опубликованные на [портале открытых данных Министерства культуры](https://opendata.mkrf.ru). Набор данных содержит: информацию о прокатных удостоверениях, сборах и государственной поддержке фильмов, информацию с сайта "КиноПоиск".

Мы получим данные:

* О прокатных удостоверениях из файла `mkrf_movies.csv`. У одного фильма может быть несколько прокатных удостоверений;
* О прокате в российских кинотеатрах из файла `mkrf_shows.csv`.

Датафрейм `mkrf_movies` содержит следующие данные:

* `title` — название фильма


* `puNumber` — номер прокатного удостоверения


* `show_start_date` — дата премьеры фильма


* `type` — тип фильма


* `film_studio` — студия-производитель


* `production_country` — страна-производитель


* `director` — режиссёр


* `producer` — продюссер


* `age_restriction` — возрастная категория


* `refundable_support` — объём возвратных средств государственной поддержки


* `nonrefundable_support` — объём невозвратных средств государственной поддержки


* `financing_source` — источник государственного финансирования


* `budget` — общий бюджет фильма. Данные в этом столбце указаны только для тех фильмов, которые получили государственную поддержку 


* `ratings` — рейтинг фильма на КиноПоиске


* `genres` — жанр фильма


Таблица `mkrf_shows` содержит следующие столбцы:

* `puNumber` — номер прокатного удостоверения


* `box_office` — сборы в рублях

**Ход исследования**

Перед исследованием потребуется обзор данных, так как отсутствует информация об их качестве. Проверим данные на наличие ошибок и исправим самые критичные ошибки на этапе предобработки.

Таким образом, исследование пройдёт в 3 этапа:

1. Обзор данных.

Ознакомимся с содержанием, структурой и характеристиками данных. Посмотрим, какие потенциальные проблемы могут быть в данных.

Проведём более тщательную проверку на наличие ошибок и оценим их влияние.

Используя обработанные данные, проведём анализ и определим тренды на рынке российского кинопроката и насколько зрителю интересны фильмы, получившие государственную поддержку.

## Обзор данных

Откроем файлы с данными:

```python
# импортируем библиотеку pandas
import pandas as pd
```

```python
# сохраним в переменную ссылку на первый датафрейм - mkrf_movies
mkrf_movies = pd.read_csv('https://code.s3.yandex.net/datasets/mkrf_movies.csv')
```

```python
# сохраним в переменную ссылку на втторой датафрейм - mkrf_shows в переменную
mkrf_shows = pd.read_csv('https://code.s3.yandex.net/datasets/mkrf_shows.csv')
```

Чтобы избежать ошибок при объединении двух датафреймов по значению столбца `puNumber`, убедимся, что в каждом из датафреймов этот столбец имеет одинаковый тип данных. 

Ознакомимся с общей информацией датафреймов:

```python
# ознакомимся с общей информацией mkrf_movies
mkrf_movies.info()
```

```python
# ознакомимся с общей информацией mkrf_shows
mkrf_shows.info()
```

В таблице `mkrf_movies` столбец `puNumber` принадлежит к типу данных **object**, в таблице `mkrf_shows` — к **int**. Приведём значения столбца в первом датафрейме к целочисленному типу: 

```python
# приведём значения столбца к int с помощью to_numeric() и astype()
# с помощью dropna() удалим значение NA, появившееся после применения to_numeric() и errors='coerce'
mkrf_movies['puNumber'] = pd.to_numeric(mkrf_movies['puNumber'], errors='coerce').dropna().astype(int)

# проверим результат
mkrf_movies.info()
```

Объединим таблицы `mkrf_movies` и `mkrf_shows`:

```python
# объединим таблицы по совпадению значений столбца puNumber 
# с помощью параметра how='outer' объединим данные из внешней общей области
# сохраним результат в mkrf
mkrf = mkrf_movies.merge(mkrf_shows, on='puNumber', how='outer')

# выведем первые 20 строк новой таблицы и проверим результат
mkrf.head(20)
```

Предварительно в таблице с данными можно заметить следующие проблемы:

* Название столбца `puNumber` в **camelCase** вместо **snake_case**.
* Вещественный тип данных в стообце `puNumber`. Обычно номер прокатного удостоверения — целое число.
* Вещественный тип данных в столбце `box_office` со сборами в рублях. В этом исследовании понадобятся значения в рублях без копеек.

## Предобработка данных

<a name="2.1"></a>
### Переименование столбцов

Проверим названия столбцов датафрейма `mkrf`:

```python
# выведем список названий столбцов 
mkrf.columns
```

В названии столбца `puNumber` заменим верблюжий регистр на змеиный:

```python
# заменим исходное название с помощью метода rename()
mkrf = mkrf.rename(columns={'puNumber':'pu_number'})

# проверим результат
mkrf.columns
```

<a name="2.2"></a>
### Изменение типов данных

Приведём столбец `show_start_date` со строковым типом данных к типу `datetime`:

```python
# преобразуем строку в datetime с помощью to_datetime()
mkrf['show_start_date'] = pd.to_datetime(mkrf['show_start_date'])

# проверим результат
mkrf['show_start_date'].dtype
```

Преобразуем строковый тип данных в столбце `ratings` в `float`:

```python
# заменим тип данных на float с помощью to_numeric()
# используем errors='coerce', чтобы избежать ошибок
mkrf['ratings'] = pd.to_numeric(mkrf['ratings'], errors='coerce')

# проверим результат
mkrf['ratings'].dtype
```

Приведём к целочисленному типу значения в столбцах `refundable_support`, `nonrefundable_support`, `budget`, `box_office`:

```python
# сохраним в переменную столбцы, в которых значения — с нулевой дробной частью
columns_to_astype = ['refundable_support', 'nonrefundable_support', 'budget']

# переведём значения в Int64, так как он может хранить NaN 
mkrf[columns_to_astype] = mkrf[columns_to_astype].astype('Int64')

# округлим значения столбца box_office с ненулевой дробной частью и переведём в Int64
mkrf['box_office'] = mkrf['box_office'].round().astype('Int64')

# проверим результат
mkrf.info()
```

Преобразуем столбец `pu_number` из вещественного в целочисленный тип данных: 

```python
# преобразуем тип в Int64, так как он может хранить NaN
mkrf['pu_number'] = mkrf['pu_number'].astype('Int64')

# проверим результат
mkrf['pu_number'].dtype
```

<a name="2.3"></a>
### Обработка пропущенных значений

Ещё раз ознакомимся с количеством пропущенных значений в каждом столбце датафрейма: 

```python
# выведем количество пропусков в каждом столбце по убыванию
mkrf.isna().sum().sort_values(ascending=False)
```

Также посмотрим на доли пропусков по столбцам c числами:

```python
# посмотрим на доли пропусков с помощью isna() и mean()
mkrf.isna().mean().sort_values(ascending=False)*100
```

* Больше всего пропусков в столбцах с денежными средствами в рублях: `refundable_support`, `nonrefundable_support` и `budget`  — **7154**, а также в `box_office` — **4328**.
* Меньше всего пропущенных значений в столбцах: `director` — **9**, `production_country` — **2**, `pu_number` — **1**.  

Заполнение пропусков не требуется в некоторых столблцах, так как не все фильмы в таблице получили государственную поддержку. Например, если заполнить пропуски медианой в столбцах `refundable_support`, `nonrefundable_support`, `budget` это будет означать, что фильмы получили поддержку от государства в определённом размере и привести к некорректным результатам исследования.  

В столбце `box_office` со сборами в рублях слишком много пропущенных значений, почти половина — **4 328** из **7 486**. Заполнение пропусков медианой исказит результаты исследования. Нежелательно заполнение числом 0, которое может означать, что фильм не вышел в прокат. 

Столбцы `genres`, `producer`, `film_studio`, `director` и `production_country`, `financing_source` содержат категориальные переменные. Пропуски в этих столбцах можно заполнить значением "неизвестно" для последующего выявления неявных дублей, более удобного анализа и восприятия:

```python
# выберем нужные столбцы и сохраним их в переменную
category_columns_to_replace = ['genres', 'producer', 'film_studio', 'director', 'production_country', 'financing_source']

# заполним пропуски с помощью fillna()
mkrf[category_columns_to_replace] = mkrf[category_columns_to_replace].fillna('неизвестно')

# проверим результат
mkrf.isna().sum()
```

Ознакомимся с информацией о фильме, у которого отсутствует номер прокатного удостоверения в столбце `pu_number`:

```python
# отфильтруем строки с пропусками
mkrf[mkrf['pu_number'].isna()]
```

Данные об этом фильме выгрузились корректно, так как на [портале открытых данных Министерства культуры](https://opendata.mkrf.ru/) в разделе ["Реестр прокатных удостоверений фильмов"](https://opendata.mkrf.ru/opendata/7705851331-register_movies) действительно отсутсвует информация о номере прокатного удостоверения для фильма "Курбан-роман. (История с жертвой)" от 2014.05.15. 

Согласно информации на портале, этот фильм получил номер прокатного удостоверения позже — 2014.09.02. В названии фильма появилась точка - "Курбан-роман. (История с жертвой).". 

Выведем информацию о фильме в датафрейме `mkrf`:

```python
# отфильтруем строки датафрейма
mkrf[mkrf['title'] == 'Курбан-роман. (История с жертвой).']
```

Номер удостоверения фильма — **111009614**, а премьера состоялась 2014.09.02. Сомнения вызывает значение **0** в столбце о сборах в рублях — `box_office`. Нулевое значение говорит о том, что фильм не вышел в прокат или была допущена ошибка при внесении данных в таблицу. Мы обработаем аномалии в этом столбце позже.

Следует удалить строку с пропущенным значением в столбце `pu_number` из датафрейма, так как данные об этом фильме трудно назвать информативными из-за большого отсутствия пропусков в разных столбцах.

```python
# удалим строку со значением NA и сохраним результат
mkrf = mkrf[mkrf['pu_number'].notna()]

# проверим результат
mkrf.isna().sum()
```

В этом случае, значение **0** соответствует действительности. Иногда не требуется возвращать средства государству. Например, если речь идёт о детских фильмах или социально значимом кино.

Несмотря на маленькие сборы, значение вполне реалистично. Неудача фильма в прокате может быть связана с плохой рекламой или слабым сценарием.

Рассчитаем межквартильный размах, чтобы наиболее точно определить верхнюю границу выбросов:

```python
# напишем функцию для расчётов с параметром - столбцом датафрейма
def emission_limit(data, column):

# рассчитаем первый квартиль - 25%
# рассчитаем второй квартиль - 75%
# рассчитаем межквартильный размах - IQR
# вычислим верхнюю границу для выбросов
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    upper_limit = Q3 + 1.5 * IQR
    return f'Верхняя граница выбросов = {upper_limit:.2f}'

# вызовем функцию и выведем результат на экран
emission_limit(mkrf, 'refundable_support')
```

Согласно расчётам, верхняя граница нормального размаха составляет — `37 500 000 рублей`. 

Отфильтруем значения в столбце:

```python
# создадим копию исходного датафрейма
good_mkrf = mkrf.copy()

# отфильтруем строки датафрейма, у которых значение в столбце выше 37_500_000
# оставим строки с NaN в столбце с помощью isna()
good_mkrf = good_mkrf[(mkrf['refundable_support'] <= 37_500_000) | good_mkrf['refundable_support'].isna()]

# проверим результат
good_mkrf['refundable_support'].sort_values(ascending=False)
```

Cравним результат предобработки, наложим гистограмму датафрейма до фильтрации — `mkrf` на гистограмму очищенных данных в `good_mkrf`:

```python
# построим гистограммы на основе исходных данных
ax = mkrf.plot(
    kind='hist',
    y='refundable_support',
    histtype='step',
    range=(0, 180_000_000),
    bins=50,
    linewidth=5,
    alpha=0.7,
    label='raw',
)
# накладываем на гистограмму после предобработки данных
good_mkrf.plot(
    kind='hist',
    y='refundable_support',
    histtype='step',
    range=(0, 180_000_000),
    bins=50,
    linewidth=4,
    alpha=0.7,
    label='filtered',
    ax=ax,
    grid=True,
    legend=True,
) 

# дадим название графику и осям
ax.set_title('Объём возвратных средств государственной поддержки')
ax.set_xlabel('Государственная поддержка в ₽')
ax.set_ylabel('Количество фильмов')
```

* После фильтрации данных распределение в столбце фатафрейма стало более концентрированным. Это позволит избежать искажений при дальнейшем анализе.

### Категоризация данных

Создадим столбец с информацией о годе проката. Выделим год из даты премьеры фильма:

```python
# отключим предупреждение SettingWithCopyWarning, так как good_mkrf это уже копия исходного mkrf
pd.options.mode.chained_assignment = None

# выделим год из даты с помощью dt.year
good_mkrf['release_year'] = good_mkrf['show_start_date'].dt.year

# проверим результат
good_mkrf.head()
```

Создадим два столбца: с именем и фамилией главного режиссёра и основным жанром фильма. В столбцы войдут первые значения из списка режиссёров и жанров.

```python
# напишем функцию для создания 2 новых столбцов 
# разобьём строки на списки элементов в выбранных столбцах, выберем первый элемент  
# вернём Series, чтобы создать 2 колонки сразу
def main_genre_and_director(row):
    main_director = row['director'].split(',')[0] 
    main_genre = row['genres'].split(',')[0]
    return pd.Series([main_director, main_genre], index=['main_director', 'main_genre'])

# вызовем функцию с помощью apply() 
good_mkrf[['main_director', 'main_genre']] = good_mkrf.apply(main_genre_and_director, axis=1)

# проверим результат
good_mkrf.tail(40)
```

Добавим столбец с долей государственной поддержки от общего бюджета фильма:

```python
# сложим возвратные и невозвратные средства гос. поддержки
# разделим сумму на бюджет
# округлим результат и сохраним в столбец state_support_share
good_mkrf['state_support_share_%'] = (
    (good_mkrf['refundable_support'] + good_mkrf['nonrefundable_support']) 
    / good_mkrf['budget'] * 100).astype(float).round(1)

# проверим результат
good_mkrf.tail(10)
```

Изменим в датафрейме столбец `age_restriction`. Сформулируем категории более кратко, для удобного обращения к элементам столбца в ходе анализа:

- «0+» - для любой зрительской аудитории — `0 - 12 лет`
- «6+» - для детей старше 6 лет — `6 - 12 лет`
- «12+» - для детей старше 12 лет — `12 - 16 лет`
- «16+» - для детей старше 16 лет — `16 - 18 лет`
- «18+» - запрещено для детей — `от 18 лет`

```python
def age_restriction(row):
    if '0' in row:
        return '0 - 6 лет'
    elif '6' in row:
        return '6 - 12 лет'
    elif '12' in row:
        return '12 - 16 лет'
    elif '16' in row:
        return '16 - 18 лет'
    elif '18' in row:
        return 'от 18 лет'

#print(age_restriction('12'))

# вызовем функцию с помощью apply
good_mkrf['age_restriction'] = good_mkrf['age_restriction'].apply(age_restriction)

# проверим результат
good_mkrf.head(10)
```

* Посмотрим, фильмы с каким возрастным ограничением собрали больше всего денег в прокате.

Посмотрим, фильмы для какой возрастной группы собрали больше всего денег в прокате независимо от года проката:

```python
# построим сводную таблицу и рассчитаем сумму сборов для каждой возрастной группы
good_mkrf.pivot_table(index='age_restriction', values='box_office', aggfunc='sum').sort_values(by='box_office', ascending=False)
```

```python
# отфильтруем фильмы с годом проката от 2015 до 2019
release_year_query = good_mkrf.query('release_year >= 2015 and release_year <= 2019')

# проверим результат
release_year_query.head()
```

Посмотрим, фильмы для какой возрастной группы собрали больше всего денег в прокате в период `с 2015 по 2019`:

```python
# построим сводную таблицу для расчётов 
release_year_query.pivot_table(index=['age_restriction'], values='box_office', aggfunc='sum').sort_values(by='box_office', ascending=False)
```

Динамика кассовых сборов по годам в зависимости от рейтингов фильмов может быть связана с изменением предпочтений аудитории, маркетинговыми стратегиями и экономическими факторами, а также с популярностью жанров.

### Исследование фильмов с государственной поддержкой

* Посмотрим, сколько средств выделяют на поддержку кино.
* Проверим, хорошо ли фильмы окупаются.

Выделим фильмы с государственной поддержкой в отдельный датафрейм:

```python
# создадим датафрейм с гос.поддержкой с помощью query()
supported_good_mkrf = good_mkrf.query('refundable_support > 0 or nonrefundable_support > 0')

# проверим результат
supported_good_mkrf.head()
```

Посмотрим, сколько средств было выделено на поддержку кино в целом:

```python
# сложим сумму возвратной и невозвратной поддержки
supported_good_mkrf['refundable_support'].sum() + supported_good_mkrf['nonrefundable_support'].sum()
```

Посмотрим, хорошо ли окупаются фильмы с государственной поддержкой:

```python
# разделим сумму сборов на общую сумму гос. поддержки
# округлим результат
(supported_good_mkrf['box_office'].sum() / (supported_good_mkrf['refundable_support'].sum() + supported_good_mkrf['nonrefundable_support'].sum())).round(2)
```

Проверим, какие фильмы с государственной поддержкой окупаются лучше всего:

```python
# создадим копию исходного датафрейма
supported_good_mkrf_copy = supported_good_mkrf.copy()

# рассчитаем окупаемость для каждого фильма и добавим новый столбец
supported_good_mkrf_copy['movie_payback'] = supported_good_mkrf_copy['box_office'] / (
                                            supported_good_mkrf_copy['refundable_support'] + 
                                            supported_good_mkrf_copy['nonrefundable_support'])
# проверим результат
supported_good_mkrf_copy.head()
```

Выведем информацию о 10 фильмах, которые окупаются лучше всего:

```python
supported_good_mkrf_copy.head(10).sort_values(by='movie_payback', ascending=False)
```
